### How to read an error before you fix it

Python's traceback is read **bottom up**. The last line names the exception and the problem; the line above it points at your code.

```
Traceback (most recent call last):
  File "<stdin>", line 2, in <module>
    print(greeting)
NameError: name 'greeting' is not defined
```

Four exceptions cover most of what you will hit this semester:

| Exception | Usually means |
|---|---|
| `NameError` | The variable was never created, or the cell that created it never ran |
| `TypeError` | You mixed types — text where a number belongs is the classic |
| `KeyError` | That column or dictionary key does not exist. Check spelling and case |
| `FileNotFoundError` | The path is wrong for *this* environment |

### Bug 2 — text where a number belongs

This should print `30`. Note that `'10' + 20` and `'10' + '20'` fail in different ways — make sure you know which one you are looking at.

````python
# TODO: fix so total is the number 30
total = 10 + 20
print(total)
````

**What was wrong:** There was a type error. The code uses a string and an int together. The goal was to add the two numbers but the + operator servers different purposes for types. For a string, + concatenates strings together. For example, '10' + '20'  gives '1020'. However, + operator for ints actually computes the sum, so 10 + 20 would give 30. In the code, python expected a string to concatenate with '10' but we had an int instead. We need to use ints to compute sum. Explaination: The fix is to change '10' to an int so the both parts on the + operator match types (int) so that they total to 30.
